```python
import matplotlib.pyplot as plt
import json
```

```python
with open('stats.json', mode='r') as json_file:
    y = json.JSONDecoder().decode(json_file.read())
    train_accuracies = y['train_accuracies']
    train_losses = y['train_losses']
    validation_accuracies = y['validation_accuracies']
    validation_losses = y['validation_losses']
    epoch_times = y['epoch_times']
```

```python
layout = [
    ['ta', 'tl'],
    ['va', 'vl'],
    ['et', 'et']
]
fig, axd = plt.subplot_mosaic(layout)

axd['ta'].plot(train_accuracies)
axd['ta'].set_ylim([0,100])
axd['tl'].plot(train_losses)
axd['tl'].set_ylim(0)
axd['va'].plot(validation_accuracies)
axd['va'].set_ylim([0,100])
axd['vl'].plot(validation_losses)
axd['vl'].set_ylim(0)
axd['et'].plot(epoch_times)
axd['et'].set_title('Times')
plt.tight_layout()
plt.show()
```

```python
plt.figure(0, figsize=[10,5])
plt.subplot(1,2,1)
plt.plot(train_losses)
plt.plot(validation_losses)
plt.legend(labels=['train','validation'])
plt.subplot(1,2,2)
plt.plot(train_accuracies)
plt.plot(validation_accuracies)
plt.legend(labels=['train','validation'])
plt.ylim([0,100])
plt.show()
```
